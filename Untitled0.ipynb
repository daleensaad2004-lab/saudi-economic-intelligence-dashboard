```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

df = pd.read_csv("Saudi_Economic_Indicators.csv")

# بدءًا من 1971 عشان قيمة 1970 الشاذة ما تخرب الرسم
df = df[df["Year"] >= 1971]

indicators = {
    "GDP_Growth": "GDP Growth (annual %)",
    "Inflation": "Inflation (CPI, annual %)",
    "Unemployment": "Unemployment (% of labor force)",
    "FDI": "FDI Net Inflows (% of GDP)",
}

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

for ax, (col, title) in zip(axes.flat, indicators.items()):
    data = df.dropna(subset=[col])
    ax.plot(data["Year"], data[col], linewidth=2)
    ax.axhline(0, color="gray", linewidth=0.8, linestyle="--")
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.set_xlabel("Year")
    ax.grid(alpha=0.3)

fig.suptitle("Saudi Arabia Economic Indicators", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.savefig("saudi_dashboard.png", dpi=200)
plt.show()

files.download("saudi_dashboard.png")
```
